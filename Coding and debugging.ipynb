```python
#1.Q1. Create a DataFrame with the following student marks and display students with marks above 80.
{'Name': ['Anu', 'Raj', 'Diya', 'Kiran'], 'Marks': [76, 88, 95, 65]}

expected Output:

    Name  Marks
1   Raj     88
2  Diya     95
```

```python
#2.Q2. Create a DataFrame and find the mean salary grouped by department.
{'Name': ['Amit', 'Sneha', 'Kunal', 'Riya'], 'Department': ['HR', 'IT', 'HR', 'IT'], 'Salary': [30000, 50000, 35000, 52000]}
Expected Output:
Department
HR    32500.0
IT    51000.0
Name: Salary, dtype: float64
```

```python
import pandas as pd

data = {'Name': ['Amit', 'Sneha', 'Kunal', 'Riya'],
        'Department': ['HR', 'IT', 'HR', 'IT'],
        'Salary': [30000, 50000, 35000, 52000]}

df = pd.DataFrame(data)
print(df.groupby('Department')['Salary'].mean())
```

```python
#3.Q3. Create a DataFrame and apply label encoding to the Gender column.
{'Name': ['Arjun', 'Leela', 'Kiran'], 'Gender': ['Male', 'Female', 'Male']}
Expected Output:
    Name  Gender  Gender_encoded
0  Arjun    Male               1
1  Leela  Female               0
2  Kiran    Male               1
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

data = {'Name': ['Arjun', 'Leela', 'Kiran'], 'Gender': ['Male', 'Female', 'Male']}
df = pd.DataFrame(data)

le = LabelEncoder()
df['Gender_encoded'] = le.fit_transform(df['Gender'])
print(df)
```

```python
#4.🔹Q4. Normalize the 'Score' column using Min-Max scaling.
{'Student': ['A', 'B', 'C'], 'Score': [55, 80, 65]}

Expected Output:
    Student  Score  Normalized_Score
0       A     55               0.0
1       B     80               1.0
2       C     65               0.4
```

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

data = {'Student': ['A', 'B', 'C'], 'Score': [55, 80, 65]}
df = pd.DataFrame(data)

scaler = MinMaxScaler()
df['Normalized_Score'] = scaler.fit_transform(df[['Score']])
print(df)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

data = {'Age': [22, 25, 47, 52, 46, 56, 23],
        'Salary': [15000, 29000, 48000, 60000, 52000, 61000, 17000],
        'Buy': [0, 0, 1, 1, 1, 1, 0]}
df = pd.DataFrame(data)

X = df[['Age', 'Salary']]
y = df['Buy']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

models = {
    "SVM": SVC(),
    "Random Forest": RandomForestClassifier(),
    "Naive Bayes": GaussianNB()
}

for name, model in models.items():
    model.fit(X_train, y_train)
    preds = model.predict(X_test)
    print(f"{name} Accuracy: {accuracy_score(y_test, preds):.2f}")
```

```python
2. Mismatched Tokenizer and Model

#	fixing the Bug
1	AutoTokenizer.from_pretrained("bert-base-uncased") with GPT-2 model and	Use "gpt2" tokenizer
2	GPT-2 has no pad token	and Set pad_token_id to avoid warning

from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("gpt2")          # Fix 1: match tokenizer to model
model = AutoModelForCausalLM.from_pretrained("gpt2")

input_text = "What is Generative AI?"
inputs = tokenizer(input_text, return_tensors="pt")
outputs = model.generate(**inputs, pad_token_id=tokenizer.eos_token_id)  # Fix 2: pad_token_id

print(tokenizer.decode(outputs[0]))
```
